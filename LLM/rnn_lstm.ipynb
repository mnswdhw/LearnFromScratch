```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
class RNN(nn.Module):

    def __init__(self, vocab_size, hidden_dim):
        super().__init__()
        self.tanh = nn.Tanh()
        self.W_h = nn.Parameter(torch.randn(hidden_dim, hidden_dim))
        self.W_x = nn.Parameter(torch.randn(vocab_size, hidden_dim))
        self.W_y = nn.Parameter(torch.randn(hidden_dim, vocab_size))

    def forward(self, X, prev_hidden):
        self.out =  self.tanh((prev_hidden @ self.W_h) + (X@self.W_x))@ self.W_y
        return F.softmax(self.out, dim = -1)
        
        
        
```

```python
mymodel = RNN(40, 32)
print(mymodel)
```
